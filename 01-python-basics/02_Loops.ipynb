while loop

```python
i = 3
while i != 0:
  print("meow")
  i = i - 1
```

```python
i = 1           # starting from 1
while i <= 3:
  print("meow")
  i = i + 1
```

```python
i = 0           # starting from 0
while i < 3:
  print("meow")
  i += 1.       # short form of increment in python
```

#for loop with list

```python
for i in [0,1,2]:  # list is defined with square brackets [ ]
  print("meow")
```

```python
for i in range(3): # range expects a number as iterable
  print("meow")
```

We can also print a string multiple times like below

```python
print("meow" * 3)
print("meow\n" * 3)       #adds an extra line in the end
print("meow\n" * 3, end="")
```

```python
while True:
  n = int(input("what's n? "))
  if n > 0:
      break

for _ in range(n):    # we can use _ instead of iterator i or n
    print("meow")
```

```python
def main():
  number = get_number()
  meow(number)

def get_number():
  while True:
    n = int(input("What's n?"))
    if n > 0:
        break
  return n

def meow(n):
  for _ in range(n):
    print("meow")

main()
```

# Dictionaries or dict

```python
students = {"Hermione": "Gryffindor",  # dict is represented using curly brackets { }
            "Harry": "Gryffindor",
            "Ron": "Gryffindor",
            "Draco": "Slytherin"}

print(students["Hermione"])
print(students["Harry"])
print(students["Ron"])
print(students["Draco"])


for student in students:
  print(student, students[student], sep=", ")
```

#dict inside list

```python
students = [
    {"name": "Hermione", "house": "Gryffindor", "patronus": "Otter"},
    {"name": "Harry", "house": "Gryffindor", "patronus": "Stag"},
    {"name": "Ron", "house": "Gryffindor", "patronus": "Jack Russell terrier"},
    {"name": "Draco", "house": "Slytherin", "patronus": None}
]

for student in students:
  print(student["name"], student["house"], student["patronus"], sep=", ")
```

#Practice

```python
# 1. Creating a list of dicts
students = [
    {"name": "Amit", "age": 20, "marks": 85},
    {"name": "Riya", "age": 21, "marks": 92},
    {"name": "Karan", "age": 19, "marks": 78},
]

print(students)
```

```python
# Accessing values
print(students[0])            # {'name': 'Amit', 'age': 20, 'marks': 85}
print(students[1]["name"])    # Riya
print(students[2]["marks"])   # 78
```

```python
# Looping through the list
for student in students:
    print(student["name"], "scored", student["marks"])
```

```python
# Adding a new dict to the list
students.append({"name": "Sneha", "age": 22, "marks": 88})
print(students)
```

```python
# Updating a value
students[0]["marks"] = 90
print(students)
```

```python
# Filtering (students with marks above 80)
toppers = [s for s in students if s["marks"] > 80]
print(toppers)
```

```python
# what is lambda function
# normal function
def add(a, b):
    return a + b

# lambda (same thing)
add = lambda a, b: a + b

print(add(3, 5))   # 8
```

Double click this block for explaination
lambda   s      :    s["marks"]
  |      |      |        |
"function" "input" "gives back" "the marks of s"

```python
# Finding the highest scorer
best = max(students, key=lambda s: s["marks"])
print(best["name"])
```

```python
# Sorting by a key
sorted_students = sorted(students, key=lambda s: s["marks"], reverse=True)

for s in sorted_students:
    print(s["name"], s["marks"])
```

```python
# Calculating the average
average = sum(s["marks"] for s in students) / len(students)
print("Average marks:", average)
```

```python
# Removing a dict from the list
students = [s for s in students if s["name"] != "Karan"]
print(students)
```
